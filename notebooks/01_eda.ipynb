# 1. Data acquisition and exploratory analysis

Dataset: **TrashNet** (Yang & Thung, Stanford CS229, 2016; MIT licence). 2,527 images, 6 classes.

This notebook downloads the images, creates a stratified 70/15/15 split, and inspects class balance and examples. The test split is locked and used only in the final evaluation.

```python
from pathlib import Path
import sys, os
ROOT = Path.cwd().parent if (Path.cwd()/'src').exists() else Path.cwd()
sys.path.insert(0, str(ROOT/'src'))
from data_pipeline import CLASS_NAMES, prepare_splits, make_datasets
print('Project root:', ROOT)
```

```python
processed = prepare_splits(ROOT/'data'/'raw', ROOT/'data'/'processed')
```

```python
import matplotlib.pyplot as plt
from PIL import Image

counts = {split: {c: len(list((processed/split/c).glob('*'))) for c in CLASS_NAMES}
          for split in ('train','val','test')}
for split, d in counts.items():
    print(split, d, 'TOTAL', sum(d.values()))

fig, axes = plt.subplots(2, 3, figsize=(10, 6))
for ax, cls in zip(axes.flat, CLASS_NAMES):
    image = sorted((processed/'train'/cls).glob('*'))[0]
    ax.imshow(Image.open(image))
    ax.set_title(f'{cls}: {counts["train"][cls]} train images')
    ax.axis('off')
plt.tight_layout(); plt.show()
```

### Observations to record in the report
- Class imbalance: `trash` is the smallest class; `paper` is the largest.
- Images were photographed on a white posterboard: easy backgrounds in training, which motivates the RealWaste cross-dataset robustness test later.
